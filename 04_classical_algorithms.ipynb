## Classification algoritms: Logistic Regression, Random Forest, SVM, XGBoost

```python
# ...existing code...
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from datasets import circle, diamond, wavy_lines  # adjust names if different in your datasets module

def split_data(data):
    train_data = data[:200]
    test_data = data[200:]
    return train_data, test_data

def unpack_data(dataset):
    X = np.array([item[0] for item in dataset])
    y = np.array([item[1] for item in dataset])
    return X, y

def decision_plot(X_test, y_test, y_pred, dataset_name, model_name):
    fig, axes = plt.subplots(1, 2, figsize=(10, 5))

    axes[0].scatter(X_test[:, 0], X_test[:, 1], c=y_test, cmap=plt.cm.Set1, s=15, edgecolors='none')
    axes[0].set_title("data")
    axes[0].set_xlabel("x1")
    axes[0].set_ylabel("x2")

    axes[1].scatter(X_test[:, 0], X_test[:, 1], c=y_pred, cmap=plt.cm.Set1, s=15, edgecolors='none')
    axes[1].set_title(r"prediction $\hat{y}$")
    axes[1].set_xlabel("x1")
    axes[1].set_ylabel("x2")

    fig.suptitle(f"{dataset_name} — {model_name}", fontsize=14)
    plt.tight_layout()
    plt.show()

# Map dataset names to generator functions
dataset_funcs = {
    'Circle': circle,
    'Diamond': diamond,
    'Wavy Lines': wavy_lines
}

for dataset_name, gen_func in dataset_funcs.items():
    print("=" * 60)
    print(f" RUNNING PIPELINE FOR DATASET: {dataset_name.upper()} ")
    print("=" * 60)

    # generate full dataset (many samples to allow split)
    data, _ = gen_func(4200)  # if your generators return only data, use: data = gen_func(4200)

    # split using the requested split_data function
    train_data, test_data = split_data(data)

    # unpack features/labels
    X_train, y_train = unpack_data(train_data)
    X_test, y_test = unpack_data(test_data)

    # scale features
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    models_to_train = {
        'Logistic Regression': LogisticRegression(max_iter=1000),
        'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42),
        'SVM': SVC(kernel='rbf'),
        'XGB': xgb.XGBClassifier(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42)
    }

    predictions = {}

    for model_name, model in models_to_train.items():
        model.fit(X_train_scaled, y_train)
        y_pred = model.predict(X_test_scaled)
        predictions[model_name] = y_pred

        print(f"\n--- {model_name} ({dataset_name}) ---")
        print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
        print("Classification Report:")
        print(classification_report(y_test, y_pred))

        # quick decision plot
        decision_plot(X_test_scaled, y_test, y_pred, dataset_name, model_name)

    # Plot confusion matrices
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle(f'Confusion Matrices for {dataset_name} Dataset', fontsize=16, fontweight='bold')

    for i, (model_name, y_pred) in enumerate(predictions.items()):
        ax = axes.flatten()[i]
        cm = confusion_matrix(y_test, y_pred)
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax, cbar=False)
        ax.set_title(f'{model_name}')
        ax.set_xlabel('Predicted')
        ax.set_ylabel('Actual')

    plt.tight_layout()
    plt.show()
# ...existing code...
```
