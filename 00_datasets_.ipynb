## Synthetic Datasets

### Circle

```python
data, (centers, radii) = circle(samples=4200)
fig = plot_circle(data, centers, radii)
plt.show()
```

```python
#Labels
print("Labels:")
print("0 <- Inside the circle")
print("1 <- Outside the circle")
print(np.array([i[1] for i in data]))
```

### Diamond

```python
    data, _ = diamond(samples=4200)
fig = plot_diamond(data)
plt.show()
```

```python
#Labels
print("Labels:")
print("0 <- Fuchsia")
print("1 <- Golden")
print("Labels:")
print(np.array([data[i][1] for i in range(len(data))]))
```

### Wavy lines

```python
data, freq = wavy_lines(samples=4200, freq=1)
fig = plot_wavy(data,freq)
plt.show()
```

```python
#Labels
print("0 <- Bellow both curves")
print("1 <- Below f1 and above f2")
print("2 <- Above f1 and below f2")
print("3 <- Above both curves")
print(np.array([i[1] for i in data]))
```
